# types of plots 

## line plot

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]
plt.plot(x,y);
plt.title("simple lint plot");
plt.xlabel("x");
plt.ylabel("y");
plt.grid(True);
```

# subplots & legends

```python
days = [1, 2, 3, 4, 5]
original_price = [10, 20, 30, 40, 50]
predicted_price = [8, 21, 25, 28, 38]
```

```python
plt.plot(days, original_price)
plt.plot(days, predicted_price)
plt.title('plot with legends')
```
